# Exploring the ClinicalTrials.gov API
First look at the ClinicalTrials.gov API v2: what a request returns and where the fields I need live.

## First request
Asking for 5 Alzheimer's trials to see the structure of the JSON. query.cond filters by condition, countTotal returns the total number of trials and pageSize sets how many trials come back per page.

```python
import requests
response = requests.get("https://clinicaltrials.gov/api/v2/studies?pageSize=5&countTotal=true&query.cond=alzheimer")
print(response.status_code)
data = response.json()
for item in data["studies"]:
    print(item["protocolSection"]["identificationModule"]["nctId"])
    print(item["protocolSection"]["identificationModule"]["briefTitle"])
    print(item["protocolSection"]["statusModule"]["overallStatus"])




```

## Where the fields live
- Total Alzheimer's trials: 4092
- phase: studies[0] → protocolSection → designModule → phases
- start date: studies[0] → protocolSection → statusModule → startDateStruct → date
- why stopped: studies[0] → protocolSection → statusModule → whyStopped (only exists for stopped trials)
