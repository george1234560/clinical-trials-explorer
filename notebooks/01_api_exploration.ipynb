```python
response = requests.get("https://clinicaltrials.gov/api/v2/studies?pageSize=5&countTotal=true&query.cond=alzheimer")
print(response.status_code)
data = response.json()
for item in data["studies"]:
    print(item["protocolSection"]["identificationModule"]["nctId"])
    print(item["protocolSection"]["identificationModule"]["briefTitle"])
    print(item["protocolSection"]["statusModule"]["overallStatus"])




```

```python
studies =[]
token = None
while True:
    response = requests.get("https://clinicaltrials.gov/api/v2/studies", params={"pageSize" :1000,"countTotal" : "true","query.cond" : "alzheimer","pageToken" : token})
    print(response.status_code)
    data = response.json()
    studies.extend(data["studies"])
    token = data.get("nextPageToken")
    if token == None:
        break
print(len(studies))
```

```python
import json
with open("../data/raw/alzheimer_trials.json","w") as f:
    json.dump(studies,f)
```
