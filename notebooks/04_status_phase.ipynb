# Conclusion
## Findings
**Status:** 53% of Alzheimer's trials are completed and 7.43% got terminated. 11.6% are UNKNOWN, meaning the sponsor stopped updating them, so some of those might have failed quietly.

**Phases:** most trials (2367) have no phase because they're observational. Of the rest, Phase 2 has the most trials (578) and there are fewer Phase 3 trials (300), which fits with a lot of drugs not making it past Phase 2.

**Termination rate per phase:** Phase 3 has the highest termination rate (20%), way above the 7.43% overall, and Phase 2 is also above average (12.11%). These are the big, expensive trials, so this could be where Alzheimer's drugs fail. But it doesn't mean the drug failed, trials also stop for money, slow recruitment or sponsor decisions. Phase 4 (5.76%) and Phase 1/2 (4.81%) are the lowest.

**Caveat:** some groups are small (Phase 2/3 has only 60 trials, Early Phase 1 has 66), so their rates can change a lot with just a few trials.
